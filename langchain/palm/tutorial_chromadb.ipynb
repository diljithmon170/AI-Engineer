```python
import chromadb

client = chromadb.Client()
collection = client.create_collection(name="my_collection")
```

```python
collection.add(
    documents=[
        "This is a sample document about AI.",
        "LangChain makes it easy to work with LLMs."
        ],
        ids=["doc1", "doc2"]
    )
```
