```python
import gensim
import pandas as pd
```

```python
df = pd.read_csv('/content/TestReviews.csv')
```

```python
df.head()
```

```python
review = df.review.apply(gensim.utils.simple_preprocess)
```

```python
review
```

training the w2c model


```python
model = gensim.models.Word2Vec(
    window = 10,
    min_count = 2,
    workers = 4
)
```

build vocabulary

```python
model.build_vocab(review, progress_per = 1000)
```

train the w2v model

```python
model.train(review, total_examples = model.corpus_count, epochs = model.epochs)
```

find similar words and similarity

```python
model.wv.most_similar('bad')
```

```python
model.wv.similarity(w1="good", w2="great")
```

```python
model.wv.similarity(w1="slow", w2="steady")
```
