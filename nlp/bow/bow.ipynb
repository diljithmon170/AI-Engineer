```python
import pandas as pd
```

```python
df = pd.read_csv("spam.csv")
df.head()
```

```python
df.Category.value_counts()
```

```python
df['spam'] = df['Category'].apply(lambda x: 1 if x == 'spam' else 0)
df.head()
```

```python
from sklearn.model_selection import train_test_split

X = df['Message']
y = df['spam']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X_train_cv = cv.fit_transform(X_train.values)
```

```python
cv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_cv, y_train)
```

```python
X_test_cv = cv.transform(X_test)
```

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_cv)
print(classification_report(y_test, y_pred))
```

```python
emails = [
    "Hey, I was wondering if you would like to grab lunch sometime this week?",
    "Congratulations! You've won a $1,000 Walmart gift card. Click here to claim your prize.",
]
```

```python
emails_cv = cv.transform(emails)
model.predict(emails_cv)
```

using pipeline

```python
from sklearn.pipeline import Pipeline

clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb_model', MultinomialNB())
])
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
print(classification_report(y_test, y_pred))
```
