```python
import numpy as np
```

```python
y_pred = np.array([0.1, 0.4, 0.35, 0.8])
y_true = np.array([0, 0, 1, 1])
```

```python
def mae(y_true, y_pred):
    total_error = 0
    for yt,yp in zip(y_true, y_pred):
        total_error +=abs(yt - yp)
    print("tot error",total_error)

    mae= total_error / len(y_pred)
    print("mae", mae)
    return mae
```

```python
mae(y_true, y_pred)
```

```python
def mse(y_true, y_pred):
    total_error = 0
    for yt,yp in zip(y_true, y_pred):
        total_error +=((yt - yp)**2)
    print("tot error",total_error)

    mse= total_error / len(y_pred)
    print("mse", mse)
    return mse
```

```python
mse(y_true, y_pred)
```

```python
np.abs(y_true - y_pred)
```

```python
np.mean(np.abs(y_true - y_pred))  # MAE
```

```python
np.mean((y_true - y_pred)**2)  # MSE
```
