```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion. i love apple. i love you")
for sent in doc.sents:
    print(sent.text)
```

```python
nlp = spacy.blank("en")
doc = nlp("captain america is a great movie. i love captain america. i love you")
for token in doc:
    print(token.text, token.idx)
```
